```python
# Bài 3: Lọc dict theo điều kiện
# Viết hàm loc_dict(d, threshold) nhận vào dict {ten: diem} và trả về dict mới chỉ chứa sinh viên có điểm > threshold.
dict = {"Minh": 9,
        "Hoang": 10,
        "Huy": 8}

def loc_dict(d, threshold):
    kq = {}
    for k, v in d.items():
        if v > threshold:
            kq[k] = v
    return kq

print(loc_dict(dict,8))
```

```python
# Bài 4: Chuyển dict thành list mô tả
# Viết hàm mo_ta_dict(d) với d là dict {id: (name, age)}, trả về list các chuỗi "ID: id - Name: name - Age: age".
dict = {1: ("Minh",22),
        2: ("Hoang",17),
        3: ("Huy", 20)}

def mo_ta_dict(d):
    list = []
    for k, v in d.items():
        list.append(f"ID: {k} - Name: {v[0]} - Age: {v[1]}")
    return list

print(mo_ta_dict(dict))
```

```python
# Bai5: Phân nhóm học sinh theo lớp
# Yêu cầu: Viết hàm phan_nhom(ds_hs) nhận vào danh sách các tuple (ten, lop) và trả về dict {lop: [ten1, ten2, ...]}.
#  Input: [('An', '10A'), ('Bình', '10A'), ('Hà', '10B')]
#     Output: {'10A': ['An', 'Bình'], '10B': ['Hà']}

dshs = [('An', '10A'), ('Bình', '10A'), ('Hà', '10B'), ('Minh', '10A')]
def phan_nhom(ds_hs):
    kq = {}
    for ds in ds_hs:
        if ds[1] in kq:
            kq[ds[1]].append(ds[0])
        else:
            kq[ds[1]] = [ds[0]]
    return kq

print(phan_nhom(dshs))
```

```python
# Bai 6  Phân loại điểm học sinh
# Yêu cầu: Viết hàm phan_loai_diem(ds) nhận danh sách tuple (ten, diem) và trả về dict {XepLoai: [ten1, ten2...]} với điều kiện:
# Giỏi: >= 8
# Khá: 6.5 - 7.9
# Trung bình: 5 - 6.4
# Yếu: < 5
dshs = [('An',10), ('Bình',4), ('Hà',6), ('Minh',7)]

def phan_loai_diem(ds:list)->dict:
    kq = {"Gioi":[],
          "Kha":[],
          "TrungBinh":[],
          "Yeu":[]}
    for ten, diem in ds:
        if diem>=8:
            kq["Gioi"].append(ten)
        if 6.5<=diem<=7.9:
            kq["Kha"].append(ten)
        if 5<=diem<=6.4:
            kq["TrungBinh"].append(ten)
        if diem<5:
            kq["Yeu"].append(ten)
    return kq
print(phan_loai_diem(dshs))
```

```python
# Bai 7 Đếm số sinh viên mỗi lớp từ dict
# Yêu cầu: Cho dict {ten: (tuoi, lop)}. Viết hàm dem_theo_lop(d) trả về dict {lop: so_luong}.
d = {"Minh":(22,"10A"),
    "Hoang":(17,"10B"),
    "Huy": (20,"10B")}

def dem_theo_lop(d: dict) -> dict:
    kq = {}
    for ten in d:
        tuoi, lop = d[ten]
        if lop in kq:
            kq[lop] += 1
        else:
            kq[lop] = 1
    return kq

print(dem_theo_lop(d))
```
